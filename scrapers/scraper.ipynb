```python
import os
import csv
import image_scraper
import music_scraper

def process_anime(title, id):
    image_folder = "images"
    music_folder = "music-clips"
    # Create folders named after the anime ID
    anime_image_folder = os.path.join(image_folder, str(id))
    anime_music_folder = os.path.join(music_folder, str(id))
    
    print(f"Processing Anime ID {id}: {title}")
    
    # Download images
    image_scraper.download_images(title, anime_image_folder)

    # Download music
    music_scraper.scrape_music(title, anime_music_folder)

if __name__ == "__main__":
    csv_file = "top_200_anime.csv"
    with open(csv_file, mode="r", encoding="utf-8") as file:
        
        reader = csv.DictReader(file)
        for row in reader:
            anime_id = row["anime_id"]
            anime_name = row["English name"]
            process_anime(title=anime_name, id=anime_id)
```
