```python
import pandas as pd
import kagglehub

# Download latest version
dbdmobile_myanimelist_dataset_path = kagglehub.dataset_download("dbdmobile/myanimelist-dataset")

print(f"Import complete. Dataset saved at {dbdmobile_myanimelist_dataset_path}")

# Load the anime dataset
anime_df = pd.read_csv(f'{dbdmobile_myanimelist_dataset_path}/anime-dataset-2023.csv')

# Convert 'Scored By' column to numeric, forcing errors to NaN
anime_df['Scored By'] = pd.to_numeric(anime_df['Scored By'], errors='coerce')

# Filter out rows where 'Scored By' column has None values
anime_df = anime_df[anime_df['Scored By'].notna()]

# Sort the anime dataset by the 'Scored By' column in descending order
anime_df = anime_df.sort_values(by='Scored By', ascending=False)

# Select the top 200 anime
top_200_anime = anime_df.head(200)

# Output the top 200 anime to a new CSV file
top_200_anime.to_csv('top_200_anime.csv', index=False)
```
